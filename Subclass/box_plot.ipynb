```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os
from pathlib import Path

def plot_comprehensive_version_comparison():
    # 1. 配置與路徑
    domains = ['F', 'G', 'H', 'I', 'J']
    output_dir = Path('../output_results/20260427')
    output_dir.mkdir(parents=True, exist_ok=True)

    # 2. 定義 8 個版本與其對應的 .base_npy 檔名模板
    versions = {
        # 舊版系列 (藍色系 - 保持不動)
        'Old No-Prompt': 'base_npy/iou_results_noprompt_domain_{}.base_npy',
        'Old Point-Prompt': 'base_npy/iou_results_point_prompt_domain_{}.base_npy',

        # 新版 unified 系列 (改為紅色系)
        'unified Baseline': 'unified_npy/iou_results_unified_baseline_domain_{}.base_npy',
        'unified MIL v1 (Point)': 'unified_npy/iou_results_unified_mil_v1_domain_{}.base_npy',
        'unified MIL v2 (ROI)': 'unified_npy/iou_results_unified_mil_v2_domain_{}.base_npy',
        'unified Q-prime (Bag)': 'unified_npy/iou_results_unified_qprime_domain_{}.base_npy',

        # SAM3 系列 (綠色系 - 保持不動)
        'SAM3 Text-Only': 'base_npy/iou_results_sam3_noprompt_domain_{}.base_npy',
        'SAM3 Text+Point': 'base_npy/iou_results_sam3_point_domain_{}.base_npy'
    }

    all_data = []

    # 3. 讀取資料並放大比例 (0-1 -> 0-100)
    for v_name, file_tmpl in versions.items():
        for d in domains:
            file_path = file_tmpl.format(d)
            if os.path.exists(file_path):
                ious = np.load(file_path)
                # 將 IoU 數值乘以 100
                for val in ious:
                    all_data.append({
                        'Domain': d,
                        'IoU (%)': val * 100,
                        'Version': v_name
                    })
            else:
                print(f"⚠️ 警告：找不到檔案 {file_path}")

    if not all_data:
        print("❌ 未偵測到任何數據，請檢查 .base_npy 檔案路徑。")
        return

    df = pd.DataFrame(all_data)

    # 4. 繪圖設定
    plt.figure(figsize=(22, 10)) # 寬度稍微再加大，確保 8 根柱子排版美觀
    sns.set_theme(style="whitegrid")

    # 5. 定義 8 個版本的專屬調色盤 (unified 改為紅色系)
    my_pal = {
        # 舊版 (藍色)
        'Old No-Prompt': '#AED6F1',           # 淺藍
        'Old Point-Prompt': '#3498DB',        # 深藍

        # unified 系列 (紅色漸層 - 區分度提高)
        'unified Baseline': '#FADBD8',        # 極淺紅 (粉)
        'unified MIL v1 (Point)': '#F1948A',  # 淺紅
        'unified MIL v2 (ROI)': '#E74C3C',    # 正紅
        'unified Q-prime (Bag)': '#943126',   # 深酒紅 (最優專家)

        # SAM3 (綠色)
        'SAM3 Text-Only': '#A9DFBF',          # 淺綠
        'SAM3 Text+Point': '#27AE60'          # 深綠
    }

    # 6. 繪製 Box Plot
    ax = sns.boxplot(
        x='Domain',
        y='IoU (%)',
        hue='Version',
        data=df,
        palette=my_pal,
        fliersize=3,
        linewidth=1.5,
        gap=0.1
    )

    # 7. 裝飾圖表
    plt.title('Comprehensive Comparison: Subclass-unified Framework vs. Baselines vs. SAM3', fontsize=22, fontweight='bold')
    plt.xlabel('Unseen Target Domains (F-J)', fontsize=16)
    plt.ylabel('Intersection over Union (IoU) %', fontsize=16)

    # 修改 Y 軸範圍為 0 到 105 (留一點空間給標籤)
    plt.ylim(0, 105)

    plt.xticks(fontsize=14)
    plt.yticks(fontsize=14)

    # 調整圖例
    plt.legend(title='Methods & Configurations', bbox_to_anchor=(1.01, 1), loc='upper left', fontsize=12, shadow=True)

    plt.tight_layout()

    # 8. 儲存
    save_path = output_dir / 'comprehensive_comparison_8bars_percentage_20260427.png'
    plt.savefig(save_path, dpi=300)
    print(f"✅ 放大比例後的紅色系整合盒鬚圖已生成：'{save_path}'")

if __name__ == "__main__":
    plot_comprehensive_version_comparison()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os
from pathlib import Path
from datetime import datetime

# ==========================================
# 專業配色定義 (使用高對比度、具學術感之深色系)
# ==========================================
PRIMARY_COLOR = '#1E40AF'  # 經典學術深藍
ERROR_BAR_COLOR = '#64748B' # 質感灰

# 實驗版本註冊表 (僅保留 Baseline 與 w/)
EXPERIMENT_REGISTRY = [
    ('Baseline',  'unified_npy/iou_results_unified_baseline_domain_{}.base_npy'),
    ('Point w/',  'unified_npy/iou_results_unified_direct_align_w_mil_point_domain_{}.base_npy'),
    ('Box w/',    'unified_npy/iou_results_unified_direct_align_w_mil_box_domain_{}.base_npy'),
    ('Mask w/',   'unified_npy/iou_results_unified_direct_align_w_mil_mask_domain_{}.base_npy'),
]

def plot_unified_comparison():
    current_date = datetime.now().strftime('%Y%m%d')
    domains = ['F', 'G', 'H', 'I', 'J']
    output_dir = Path(f'../output_results/{current_date}')
    output_dir.mkdir(parents=True, exist_ok=True)

    all_data = []

    for name, path_tmpl in EXPERIMENT_REGISTRY:
        # 將名稱轉換為簡潔的 X 軸標籤
        prompt_map = {
            'Baseline': 'Baseline',
            'Point w/': 'Point Prompt',
            'Box w/': 'Box Prompt',
            'Mask w/': 'Mask Prompt'
        }
        prompt_label = prompt_map[name]

        for d in domains:
            file_path = path_tmpl.format(d)
            if os.path.exists(file_path):
                ious = np.load(file_path)
                for val in ious:
                    all_data.append({
                        'Domain': d,
                        'IoU (%)': val * 100,
                        'Prompt': prompt_label
                    })
            else:
                print(f"⚠️ 警告：找不到 {file_path}")

    if not all_data:
        print("❌ 沒有讀取到任何資料！")
        return

    df = pd.DataFrame(all_data)

    # 使用 white 風格 (無背景網格，標準科學圖表底色)
    sns.set_theme(style="white")

    # 定義精確度遞增的 X 軸順序
    prompt_order = ['Baseline', 'Point Prompt', 'Box Prompt', 'Mask Prompt']

    for d in domains:
        df_domain = df[df['Domain'] == d].copy()
        if df_domain.empty:
            continue

        # 設定黃金比例 8:5 的畫布大小
        fig, ax = plt.subplots(figsize=(8, 5))

        # 強制指定順序
        df_domain['Prompt'] = pd.Categorical(df_domain['Prompt'], categories=prompt_order, ordered=True)
        df_domain = df_domain.sort_values('Prompt')

        # 繪製核心折線與誤差棒 (errorbar='sd' 代表標準差)
        sns.lineplot(
            x='Prompt',
            y='IoU (%)',
            data=df_domain,
            marker='o',
            markersize=10,
            markerfacecolor='white',      # 白色內圈錨點
            markeredgewidth=3,            # 錨點邊框加粗
            markeredgecolor=PRIMARY_COLOR, # 錨點邊框顏色
            linewidth=3,                  # 核心折線加粗
            color=PRIMARY_COLOR,
            errorbar='sd',                # 改用科學圖表常用的 Standard Deviation
            err_style='bars',             # 將陰影改為標準誤差棒
            err_kws={'capsize': 5, 'elinewidth': 1.5, 'ecolor': ERROR_BAR_COLOR}, # 誤差棒樣式
            ax=ax
        )

        # ==========================================
        # 專業圖表細節微調 (Aesthetics)
        # ==========================================
        # 標題與軸標籤
        ax.set_title(f'Env {d}', fontsize=16, fontweight='bold', pad=15)
        ax.set_xlabel('Prompt Granularity (Increasing Precision)', fontsize=12, fontweight='bold', labelpad=10)
        ax.set_ylabel('IoU Score (%)', fontsize=12, fontweight='bold', labelpad=10)

        # 軸刻度調整
        ax.tick_params(axis='both', labelsize=11)
        ax.set_ylim(0, 105)

        # 移除上方與右方框線 (經典學術圖表風格)
        sns.despine(top=True, right=True)

        # 僅保留水平的淡淡參考線，避免干擾折線閱讀
        ax.yaxis.grid(True, linestyle='--', alpha=0.5, color='#CBD5E1')
        ax.xaxis.grid(False)

        # 儲存與關閉
        plt.tight_layout()
        save_path = output_dir / f"MIL_trend_Env_{d}_{current_date}.png"
        plt.savefig(save_path, dpi=300, bbox_inches='tight')
        plt.close()
        print(f"✅ 專業圖表已生成：{save_path}")

if __name__ == "__main__":
    plot_unified_comparison()
```

```python
from pathlib import Path

def batch_rename_files(folder_path, old_text, new_text, test_mode=True):
    """
    批次修改資料夾內的檔案名稱。

    :param folder_path: 檔案所在的資料夾路徑 (例如: "Subclass/unified_npy")
    :param old_text: 檔名中想要被替換掉的舊字串
    :param new_text: 想要替換成的新字串
    :param test_mode: True 代表只印出結果不實際改名，False 代表正式改名
    """
    p = Path(folder_path)
    if not p.exists():
        print(f"❌ 找不到資料夾: {folder_path}")
        return

    # 找出所有開頭是 iou_results_ 的檔案 (包含 .base_npy 和 .json)
    target_files = list(p.glob("iou_results_*"))

    count = 0
    print(f"{'='*15} {'測試模式 (不會改名)' if test_mode else '正式執行 (檔案已改名)'} {'='*15}")

    for file_path in target_files:
        old_name = file_path.name

        # 檢查檔名是否包含我們要替換的目標字串
        if old_text in old_name:
            new_name = old_name.replace(old_text, new_text)
            new_file_path = file_path.with_name(new_name)

            if not test_mode:
                file_path.rename(new_file_path)

            print(f"🔄 {old_name}  -->  {new_name}")
            count += 1

    print(f"{'='*50}")
    print(f"✅ 共處理了 {count} 個檔案。")


# ================= 使用設定區 =================

# 1. 設定你的 NPY 與 JSON 存檔路徑
FOLDER = "unified_npy"

# 2. 設定你要替換的字串 (根據你上一篇的 log，舊名字應該是下面這兩個其中之一)
# 例如把 "unified_linear_Vb_w" 換成 "mil_direct_align"
OLD_STR = "linear_Vb_w"
NEW_STR = "alpha_solver_w"

# 3. 執行改名 (強烈建議先保持 test_mode=True 跑一次看結果)
#batch_rename_files(FOLDER, OLD_STR, NEW_STR, test_mode=True)

# 4. 確認無誤後，把上面那行註解掉，改成下面這行正式執行：
batch_rename_files(FOLDER, OLD_STR, NEW_STR, test_mode=False)
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os
from pathlib import Path
from datetime import datetime

# ==========================================
# 顏色庫定義
# ==========================================
COLOR_POOL = {
    'BLUE': ['#85C1E9'],
    'RED': ['#F5B7B1'],
    'GREEN': ['#A9DFBF'],
    'PURPLE': ['#D7BDE2'],
    'ORANGE': ['#FAD7A1']
}

def plot_unified_comparison(file_name,w,h):
    current_date = datetime.now().strftime('%Y%m%d')
    domains = ['F', 'G', 'H', 'I', 'J']
    FOLDS_NUM = 5  # 宣告跑 5 個 Fold

    # 輸出圖表路徑
    output_dir = Path(f'../output_results/{current_date}')
    output_dir.mkdir(parents=True, exist_ok=True)

    all_data = []
    version_colors = {}
    group_counters = {k: 0 for k in COLOR_POOL.keys()}

    for name, path_tmpl, group in EXPERIMENT_REGISTRY:
        color_idx = group_counters[group]
        version_colors[name] = COLOR_POOL[group][min(color_idx, len(COLOR_POOL[group])-1)]
        group_counters[group] += 1

        for d in domains:
            # 🔴 核心改動：加入 Fold 迴圈，把 5 個檔案的 IoU 收集到同一個 all_data 裡面
            for fold in range(FOLDS_NUM):
                # 使用 .format(fold, d) 分別填入 fold_idx 與 domain
                file_path = path_tmpl.format(fold, d)

                if os.path.exists(file_path):
                    ious = np.load(file_path)
                    for val in ious:
                        all_data.append({'Env': d, 'IoU (%)': val * 100, 'Version': name})
                else:
                    print(f"⚠️ 警告：找不到 {file_path}")

    if not all_data:
        print("❌ 沒有讀取到任何資料！")
        return

    df = pd.DataFrame(all_data)
    sns.set_theme(style="whitegrid")

    for d in domains:
        df_domain = df[df['Env'] == d]
        if df_domain.empty:
            continue

        plt.figure(figsize=(w, h))
        order = [name for name, _, _ in EXPERIMENT_REGISTRY]

        ax = sns.boxplot(
            x='Version',
            y='IoU (%)',
            hue='Version',
            data=df_domain,
            order=order,
            palette=version_colors,
            legend=False,
            width=0.4,
            fliersize=3,
            linewidth=1
        )
        # 設定 x 軸與 y 軸的刻度字體大小
        ax.tick_params(axis='both', which='major', labelsize=16)
        plt.title(f'Env. {d}', fontsize=20, fontweight='bold')
        plt.ylim(0, 105)
        plt.xlabel('')
        plt.ylabel('IoU Score (%)', fontsize=14)

        plt.xticks(
            rotation=25,
            ha='right',
            fontsize=23
        )

        plt.tight_layout()
        save_path = output_dir / f"{file_name}comparison_Env_{d}_{current_date}.png"
        plt.savefig(save_path, dpi=300, bbox_inches='tight')
        plt.close()
        print(f"✅ 圖表已生成：{save_path}")
```

### F-J 5 fold實驗 包含Precision

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os
from pathlib import Path
from datetime import datetime
from matplotlib.patches import Patch

# ==========================================
# 顏色庫定義 (Tuple 格式 -> 第一項淺色代表 IoU，第二項深色代表 Precision)
# ==========================================
COLOR_POOL = {
    'BLUE': ('#85C1E9', '#2E86C1'),
    'RED': ('#F5B7B1', '#CB4335'),
    'GREEN': ('#A9DFBF', '#28B463'),
    'PURPLE': ('#D7BDE2', '#884EA0'),
    'ORANGE': ('#FAD7A1', '#D68910')
}

def plot_unified_comparison(file_name, w, h):
    current_date = datetime.now().strftime('%Y%m%d')
    domains = ['F', 'G', 'H', 'I', 'J']
    FOLDS_NUM = 5

    output_dir = Path(f'../output_results/{current_date}')
    output_dir.mkdir(parents=True, exist_ok=True)

    all_data = []

    # 建立一個 mapping 方便稍後塗色與產生圖例
    version_group_map = {name: group for name, _, group in EXPERIMENT_REGISTRY}

    for name, path_tmpl, group in EXPERIMENT_REGISTRY:
        for d in domains:
            for fold in range(FOLDS_NUM):
                file_path = path_tmpl.format(fold, d)

                if os.path.exists(file_path):
                    metrics = np.load(file_path)

                    # 處理舊版只有 IoU (1D) 的資料
                    if metrics.ndim == 1:
                        for val in metrics:
                            all_data.append({'Env': d, 'Score (%)': val * 100, 'Metric': 'IoU', 'Version': name})
                            all_data.append({'Env': d, 'Score (%)': np.nan, 'Metric': 'Precision', 'Version': name})

                    # 處理新版 (N, 2) 的資料
                    elif metrics.ndim == 2 and metrics.shape[1] == 2:
                        for val in metrics:
                            all_data.append({'Env': d, 'Score (%)': val[0] * 100, 'Metric': 'IoU', 'Version': name})
                            all_data.append({'Env': d, 'Score (%)': val[1] * 100, 'Metric': 'Precision', 'Version': name})
                else:
                    print(f"⚠️ 警告：找不到 {file_path}")

    if not all_data:
        print("❌ 沒有讀取到任何資料！")
        return

    df = pd.DataFrame(all_data)
    sns.set_theme(style="whitegrid")

    for d in domains:
        df_domain = df[df['Env'] == d]
        if df_domain.empty:
            continue

        plt.figure(figsize=(w, h))
        order = [name for name, _, _ in EXPERIMENT_REGISTRY if name in df_domain['Version'].unique()]

        # 繪製底圖，並關閉原生的 legend 避免雜亂
        ax = sns.boxplot(
            x='Version',
            y='Score (%)',
            hue='Metric',
            data=df_domain,
            order=order,
            palette={'IoU': '#FFFFFF', 'Precision': '#CCCCCC'},
            legend=False,
            width=0.7,
            fliersize=3,
            linewidth=1.2
        )

        # ==========================================
        # 🎨 後處理塗色
        # ==========================================
        boxes = [p for p in ax.patches if type(p).__name__ in ['PathPatch', 'Rectangle']]
        num_versions = len(order)

        for i, box in enumerate(boxes):
            # 前半段為 IoU (淺色)，後半段為 Precision (深色)
            if i < num_versions:
                metric_idx = 0
                v_name = order[i]
            else:
                metric_idx = 1
                v_name = order[i - num_versions]

            group = version_group_map[v_name]
            color = COLOR_POOL[group][metric_idx]

            box.set_facecolor(color)
            box.set_edgecolor('#333333')
            box.set_alpha(0.9)

        # ==========================================
        # 🏷️ 動態生成簡潔圖例 (只跟著顏色群組數量走)
        # ==========================================
        used_groups = []
        for name in order:
            if version_group_map[name] not in used_groups:
                used_groups.append(version_group_map[name])

        legend_elements = []
        # 第 1 列：淺色區塊 (IoU)
        for group in used_groups:
            light_c, _ = COLOR_POOL[group]
            legend_elements.append(Patch(facecolor=light_c, edgecolor='#333333', label='IoU'))

        # 第 2 列：深色區塊 (Precision)
        for group in used_groups:
            _, dark_c = COLOR_POOL[group]
            legend_elements.append(Patch(facecolor=dark_c, edgecolor='#333333', label='Precision'))

        # 🔴 核心改動：使用 bbox_to_anchor 將圖例推到圖表外側上方
        ax.legend(
            handles=legend_elements,
            loc='lower right',             # 對齊圖例的右下角
            bbox_to_anchor=(1.0, 1.02),    # 放置在圖表座標的 (X=1.0 最右, Y=1.02 最頂部再往上一點)
            fontsize=20,
            framealpha=0.95,
            ncol=len(used_groups),         # 保持兩排對齊
            borderaxespad=0.               # 移除邊界多餘留白
        )

        ax.tick_params(axis='both', which='major', labelsize=20)
        plt.title(f'Env. {d}', fontsize=24, fontweight='bold')
        plt.ylim(0, 105)
        plt.xlabel('')
        plt.ylabel('Score (%)', fontsize=18)

        plt.xticks(rotation=25, ha='right', fontsize=24)

        plt.tight_layout()
        save_path = output_dir / f"{file_name}comparison_Env_{d}_{current_date}.png"
        plt.savefig(save_path, dpi=300, bbox_inches='tight')
        plt.close()
        print(f"✅ 圖表已生成：{save_path}")
```

### K-L 實驗

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os
from pathlib import Path
from datetime import datetime
from matplotlib.patches import Patch

# ==========================================
# 顏色庫 (淺色=IoU, 深色=Precision)
# ==========================================
COLOR_POOL = {
    'BLUE':   ('#85C1E9', '#2E86C1'),
    'RED':    ('#F5B7B1', '#CB4335'),
    'GREEN':  ('#A9DFBF', '#28B463'),
    'PURPLE': ('#D7BDE2', '#884EA0'),
    'ORANGE': ('#FAD7A1', '#D68910'),
}

def plot_case_study_KL(file_prefix='case_study_', w=10, h=6):
    current_date = datetime.now().strftime('%Y%m%d')
    domains = ['K', 'L']

    output_dir = Path(f'../output_results/{current_date}')
    output_dir.mkdir(parents=True, exist_ok=True)

    version_group_map = {name: group for name, _, group in EXPERIMENT_REGISTRY}

    all_data = []

    for name, path_tmpl, group in EXPERIMENT_REGISTRY:
        for d in domains:
            file_path = path_tmpl.format(d)

            if os.path.exists(file_path):
                metrics = np.load(file_path)   # shape: (trials * query_shots, 2)

                if metrics.ndim == 1:
                    for val in metrics:
                        all_data.append({'Env': d, 'Score (%)': val * 100,  'Metric': 'IoU',       'Version': name})
                        all_data.append({'Env': d, 'Score (%)': np.nan,     'Metric': 'Precision', 'Version': name})

                elif metrics.ndim == 2 and metrics.shape[1] == 2:
                    for val in metrics:
                        all_data.append({'Env': d, 'Score (%)': val[0] * 100, 'Metric': 'IoU',       'Version': name})
                        all_data.append({'Env': d, 'Score (%)': val[1] * 100, 'Metric': 'Precision', 'Version': name})
            else:
                print(f"⚠️  找不到: {file_path}")

    if not all_data:
        print("❌ 沒有讀取到任何資料！請確認 base_npy 路徑。")
        return

    df = pd.DataFrame(all_data)
    sns.set_theme(style="whitegrid")

    for d in domains:
        df_domain = df[df['Env'] == d]
        if df_domain.empty:
            print(f"⚠️  場域 {d} 無資料，略過。")
            continue

        order = [name for name, _, _ in EXPERIMENT_REGISTRY if name in df_domain['Version'].unique()]

        plt.figure(figsize=(w, h))
        ax = sns.boxplot(
            x='Version', y='Score (%)', hue='Metric',
            data=df_domain, order=order,
            palette={'IoU': '#FFFFFF', 'Precision': '#CCCCCC'},
            legend=False, width=0.7, fliersize=3, linewidth=1.2,
        )

        # ── 後處理塗色 ───────────────────────────────────
        boxes = [p for p in ax.patches if type(p).__name__ in ['PathPatch', 'Rectangle']]
        num_versions = len(order)

        for i, box in enumerate(boxes):
            metric_idx = 0 if i < num_versions else 1
            v_name = order[i] if i < num_versions else order[i - num_versions]
            color = COLOR_POOL[version_group_map[v_name]][metric_idx]
            box.set_facecolor(color)
            box.set_edgecolor('#333333')
            box.set_alpha(0.9)

        # ── 圖例 ─────────────────────────────────────────
        used_groups = []
        for name in order:
            g = version_group_map[name]
            if g not in used_groups:
                used_groups.append(g)

        legend_elements = (
            [Patch(facecolor=COLOR_POOL[g][0], edgecolor='#333333', label='IoU')       for g in used_groups] +
            [Patch(facecolor=COLOR_POOL[g][1], edgecolor='#333333', label='Precision') for g in used_groups]
        )

        ax.legend(
            handles=legend_elements,
            loc='lower right',
            bbox_to_anchor=(1.0, 1.02),
            fontsize=20, framealpha=0.95,
            ncol=len(used_groups),
            borderaxespad=0.,
        )

        ax.tick_params(axis='both', which='major', labelsize=24)
        plt.title(f'Env. {d}(5-shot)', fontsize=24, fontweight='bold')
        plt.ylim(0, 105)
        plt.xlabel('')
        plt.ylabel('Score (%)', fontsize=24)
        plt.xticks(rotation=25, ha='right', fontsize=24)
        plt.tight_layout()

        save_path = output_dir / f"{file_prefix}Env_{d}_{current_date}.png"
        plt.savefig(save_path, dpi=300, bbox_inches='tight')
        plt.close()
        print(f"✅ 圖表已儲存: {save_path}")
```
